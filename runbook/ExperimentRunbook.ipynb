```python

#exp_set_1AB_side_22-05-10to23-11-02_AtoKD--200-OHLC.csv
def experiment(model_name, learning_rate, datafilename, data_columns,  shape_start, shape_end, shape_power):

  # this is the experiment cell for the shaped ETS former experiment
  #model_name="Transformer"
  !export CUDA_VISIBLE_DEVICES=4

  #%cd Time-Series-Library-fork
  !pwd

  #t="$(date +""%Y%m%d-%H--%M"")"
  #%mkdir /content/drive/MyDrive/UnattendedResults/$t

  !python -u run.py \
    --task_name long_term_forecast \
    --is_training 1 \
    --root_path dataset/disertation_data/ \
    --data_path $datafilename \
    --model_id Exchange_96_96 \
    --model $model_name \
    --data custom_multi \
    --features M \
    --seq_len 96 \
    --label_len 48 \
    --pred_len 96 \
    --e_layers 2 \
    --d_layers 2 \
    --factor 3 \
    --enc_in $data_columns \
    --dec_in $data_columns \
    --learning_rate $learning_rate \
    --attn_shape_start $shape_start \
    --attn_shape_end $shape_end \
    --attn_shape_power $shape_power \
    --c_out $data_columns \
    --des 'Exp' \
    --itr 1

  with open("params.txt", "a") as f:
    f.write(f"run parameters model:{model_name} \nlearning_rate:{learning_rate}\n")
    f.write(f"run parameters datafilename:{datafilename} \data_columns:{data_columns}\n")
    f.write(f"run parameters shape_start:{shape_start} \shape_end:{shape_end} shape_power:{shape_power}\n")


  %mv ./params.txt ../drive/MyDrive/UnattendedResults/$folder/"$(date +""%m-%d_%H-%M+1H-params.txt"")"
  %mv ./result_long_term_forecast.txt ../drive/MyDrive/UnattendedResults/$folder/"$(date +""%m-%d_%H-%M+1H-result_long_term_forecast--ETS-gettingstarted--.txt"")"
  %mv ./complete_experiment_long_term_forecast.txt ../drive/MyDrive/UnattendedResults/$folder/"$(date +""%m-%d_%H-%M+1H--ETS_getting_started_complete_experiment_long_term_forecast.txt"")"
  !find -name '*bias*' | zip biases-ets.zip -@
  !rm  *bias_history*
  !find -name 'self.fcount_*' | zip head_imgs-ets.zip -@
  !rm  *self.fcount_*
  %mv ./biases-ets.zip ../drive/MyDrive/UnattendedResults/$folder/"$(date +""%m-%d_%H-%M+1H--biases-ets.zip"")"
  %mv ./head_imgs-ets.zip ../drive/MyDrive/UnattendedResults/$folder/"$(date +""%m-%d_%H-%M+1H--head_imgs-ets.zip"")"
  !zip -r resultpdfs.zip ./test_results
  %mv ./resultpdfs.zip ../drive/MyDrive/UnattendedResults/$folder/"$(date +""%m-%d_%H-%M+1H--resultpdfs-ets.zip"")"
  !zip -r resultnpys.zip ./results
  %mv ./resultnpys.zip ../drive/MyDrive/UnattendedResults/$folder/"$(date +""%m-%d_%H-%M+1H--resultnpys-ets.zip"")"
```
